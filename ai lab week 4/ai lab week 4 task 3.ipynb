```python
g={"S":[("A",1),("B",4)],"A":[("C",2)],"B":[("G",5)],"C":[("G",3)],"G":[]}
h={"S":5,"A":4,"B":4,"C":10,"G":0}
import heapq
def astar(graph,h,start,goal):
    pq=[(h[start],0,start,[start])]; best={start:0}
    tracenode=[]
    countnode=0
    while pq:
        f,g,node,path=heapq.heappop(pq)
        tracenode.append(node)
        countnode+=1
        if node==goal: return g,path,tracenode,countnode
        if g!=best[node]: continue
        for i,w in graph.get(node,[]):
            ng=g+w
            if ng < best.get(i,float("inf")):
                best[i]=ng
                heapq.heappush(pq,(ng+h[i],ng,i,path+[i]))
def ucs(graph,start,goal):
    pq=[(0,start,[start])]; best={start:0}
    tracenode=[]
    countnode=0
    while pq:
        g,node,path=heapq.heappop(pq)
        tracenode.append(node)
        countnode+=1
        if node==goal: return g,path,tracenode,countnode
        if g!=best[node]: continue
        for i,w in graph.get(node,[]):
            ng=g+w
            if ng < best.get(i,float("inf")):
                best[i]=ng
                heapq.heappush(pq,(ng,i,path+[i]))
print(astar(g,h,"S","G"))
print(ucs(g,"S","G"))
```

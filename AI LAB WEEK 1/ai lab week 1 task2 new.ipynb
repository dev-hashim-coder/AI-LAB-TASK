```python
class rulebased:
    def __init__(self,score,doucment,attendance,marks):
        self.score = score
        self.doucment = doucment
        self.attendance = attendance
        self.marks = marks
    def show(self):
        if (self.score >= 70 and self.doucment =="complete" and self.attendance >= 75 and self.marks >= 60):
            print("pass")
        else:
            print("fail")
for i in range(1,5):
    print(i)
    score=int(input("enter score:"))
    doucment=input("enter doucment:")
    attendance=int(input("enter attendance:"))
    marks=float(input("enter marks:"))
    obj = rulebased(score,doucment,attendance,marks)
    obj.show()
    print("")
    if score<70:
        expected="fail"
        reason="score is less than 70"
    elif doucment!="complete":
        expected="fail"
        reason="document incomplete"
    elif attendance<75:
        expected="fail"
        reason="attendance is less than 75"
    elif marks<60:
        expected="fail"
        reason="marks less than 60"
    else:
        expected = "pass"
        reason="condition satisfy"
    print(expected)
    print(reason)
```

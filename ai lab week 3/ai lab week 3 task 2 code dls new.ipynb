```python
graph = {"a":["b","c"],
         "b":["d","e"],
         "c":["f"],
         "d":[],
         "e":[],
         "f":["g"],
         "g":[]
        }
def dls(node,goal,path,limit):
    if node==goal:
        return "found", path
    if limit==0:
        if graph[node]:
           return "cutoff", None
        return "failure", None
    for i in graph[node]:
        if i in path:
            continue
        status,Path= dls(i,goal,path+[i],limit-1)
        if status=="found":
            return "found",Path
        if status=="cutoff":
            return "cutoff",None
    return "failure",None
for i in range(1,5):
    status,path= dls("a","g",["a"],i)
    print(i,status,path)
```

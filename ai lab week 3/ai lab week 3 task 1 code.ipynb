```python
from collections import deque
graph = { "a":["b","c"],
          "b":["d","e"],
          "c":["f"],
          "d":[],
          "e":[],
          "f":["g"],
          "g":[],
}
def bfs(a,g):
    visited = []
    queue = deque([[a]])
    while queue:
        print(queue,visited)
        temp = queue.popleft()
        node = temp[-1] 
        if node not in visited:
           visited.append(node)
        if node==g:
            return temp ,visited
        for child in graph[node]:
            if child not in visited:
                path=temp+[child]
                queue.append(path)
    return None, visited
def dfs(a,g):
    visited=[]
    stack = [[a]]
    while stack:
        print(stack,visited)
        temp = stack.pop()
        node = temp[-1]
        if node not in visited:
           visited.append(node)
        if node==g:
            return temp,visited
        for child in graph[node][::-1]:
            if child not in visited:
                 path = temp+ [child]
                 stack.append(path)
    return None, visited
bfsvisited, bfsqueu= bfs("a", "g")
dfsvisited, dfsstack= dfs("a", "g")
print(bfsvisited)
print(bfsqueu)
print(dfsvisited)
print( dfsstack)
```
